```python
#RNN (Recurrent Neural Network) is a type of neural network that is used to process sequential data by remembering information from previous steps.
#Examples: Text prediction, speech recognition, language translation and time-series forecasting.


    
# 1. Import NumPy
import numpy as np

```

```python
# NumPy is used for mathematical and matrix calculations
# seed(42) ensures the same random values every time
np.random.seed(42)

```

```python

# 2. Define RNN sizes
input_size = 1
hidden_size = 4
output_size = 1
```

```python

# 3. Initialize weights
# randn() generates random numbers from a normal distribution
w_xh = np.random.randn(hidden_size, input_size)  # Input to hidden
w_hh = np.random.randn(hidden_size, hidden_size)  # Previous hidden to current hidden
w_hy = np.random.randn(output_size, hidden_size)  # Hidden to output

# zeros() creates bias values initialized to zero
b_h = np.zeros((hidden_size, 1))
b_y = np.zeros((output_size, 1))

print("RNN parameters initialized")
```

```python
# 4. Create input sequence
sequence = [1, 2, 3, 4, 5]

# Initialize hidden state with zeros
h = np.zeros((hidden_size, 1))

print("Input sequence:", sequence)
print("Initial hidden state:\n", h)
```

```python

# 5. Process the input sequence
print("\nRecurrent Neural Network")

# enumerate() gives both the time step and input value
for t, value in enumerate(sequence, start=1):

    # array() converts the input value into a NumPy array
    x = np.array([[value]], dtype=float)

    # dot() performs matrix multiplication
    # tanh() applies activation and updates the hidden state
    h = np.tanh(
        np.dot(w_xh, x) +
        np.dot(w_hh, h) +
        b_h
    )

    # Calculate output using the hidden state
    y = np.dot(w_hy, h) + b_y

    # flatten() converts the result into a simple 1D array for printing
    print("\nTime step:", t)
    print("Input:", value)
    print("Hidden state:", h.flatten())
    print("Output:", y.flatten())
```

```python
# 6. Display final results
print("\nFinal hidden state:")
print(h)

print("\nFinal output:")
print(y)

print("\nRNN Implementation Completed Successfully!")
```

```python
#w_xh – Input weights.



#w_hh – Recurrent weights.



#w_hy – Output weights.



#h – Stores previous information.



#tanh – Activation function.



#y – Final output at each time step.
```
