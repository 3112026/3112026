```python
#CBOW (Continuous Bag of Words) is a Word2Vec model used in Natural Language Processing (NLP). It predicts a center word using the surrounding words in a sentence.
#Example
#Sentence: "I love machine learning"

#If the center word is love:
#- Input (Context words): I, machine
#- Output (Target word): love
```

```python
# 1. Import required libraries
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Embedding, GlobalAveragePooling1D, Dense
```

```python
# 2. Data Preparation
# Sample sentences for training
sentences = [
    "I love machine learning",
    "I love deep learning",
    "machine learning is interesting",
    "deep learning is powerful"
]
```

```python
# Create tokenizer to convert words into numbers
tokenizer = Tokenizer()

# Create vocabulary and assign a number to each word
tokenizer.fit_on_texts(sentences)

# Convert sentences into sequences of numbers
sequences = tokenizer.texts_to_sequences(sentences)

# Find total number of unique words
vocab_size = len(tokenizer.word_index) + 1

print("Vocabulary:", tokenizer.word_index)
```

```python
# 3. Generate Training Data
contexts = []
targets = []


# Take surrounding words as input and center word as output
for sequence in sequences:
    for i in range(1, len(sequence)-1):
        contexts.append([sequence[i-1], sequence[i+1]])
        targets.append(sequence[i])

# Convert lists into NumPy arrays
X = np.array(contexts)
y = np.array(targets)
```

```python
# 4. Build the CBOW Model
model = Sequential([
    Input(shape=(2,)),                    # Two context words as input
    Embedding(vocab_size, 10),            # Convert words into vectors
    GlobalAveragePooling1D(),             # Calculate average of word vectors
    Dense(vocab_size, activation='softmax') # Predict center word
])

#GlobalAveragePooling1D() is used for sequential data, such as text and word embeddings.
#GlobalAveragePooling2D() is used for 2D feature maps, commonly produced by CNNs processing images.
```

```python
# 5. Compile the Model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# 6. Train the Model
model.fit(X, y, epochs=50)

```

```python
# 7. Output
# Predict the center word using the first context
prediction = model.predict(X[:1])

# Find the word with the highest probability
predicted_word = np.argmax(prediction)

# Convert predicted number back into a word
print("Predicted Word:", tokenizer.index_word[predicted_word])
# Display actual center word
print("Actual Word:", tokenizer.index_word[y[0]])
```
