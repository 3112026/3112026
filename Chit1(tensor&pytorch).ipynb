```python

# Import required libraries
import tensorflow as tf
import numpy as np

# Create a 2D tensor using constant values
# tf.constant() creates a fixed tensor
x = tf.constant([[1., 2., 3.],
                 [4., 5., 6.]])

# Display tensor, shape and data type
print("Tensor:")
print(x)
print("Shape:", x.shape)  # Shows rows and columns
print("Data Type:", x.dtype)  # Shows data type

# Addition of tensors
print("\nAddition:")
print(x + x)

# Subtraction of tensors
print("\nSubtraction:")
print(x - x)

# Multiply tensor by a number
print("\nMultiplication:")
print(5 * x)

# Matrix multiplication with transpose
# tf.transpose() changes rows into columns
print("\nMatrix Multiplication:")
print(x @ tf.transpose(x))

# Concatenate joins multiple tensors
# axis=0 joins tensors row-wise
result = tf.concat([x, x, x], axis=0)
print("\nConcatenated Tensor:")
print(result)

# Softmax converts values into probabilities
# axis=-1 applies softmax along the last dimension
result = tf.nn.softmax(x, axis=-1)
print("\nSoftmax:")
print(result)

# Create two constant values and add them
a = tf.constant(15)
b = tf.constant(20)
print("\nAddition of Constants:", a + b)

# Generate 100 random values using NumPy
# astype() converts values into float32
x = np.random.rand(100).astype(np.float32)
print("\nRandom Values:")
print(x)

# Create trainable weight and bias variables
# tf.Variable() creates values that can be updated during training
W = tf.Variable(tf.random.normal([1]))
b = tf.Variable(tf.zeros([1]))

print("\nWeight:", W)
print("Bias:", b)

print("\nTensorFlow Implementation Completed!")
```

```python

# Import required libraries
import torch
import numpy as np

# Create a Python list
data = [[1, 2], [3, 4]]

# Convert list into a PyTorch tensor
# torch.tensor() creates a tensor from data
x_data = torch.tensor(data)

print("Tensor:")
print(x_data)
print("Data Type:", type(x_data))

# Convert NumPy array into PyTorch tensor
np_array = np.array(data)
x_np = torch.from_numpy(np_array)

print("\nNumPy to PyTorch Tensor:")
print(x_np)
print(type(x_np))

# Create a tensor containing ones
# ones_like() creates a tensor with the same shape as x_data
x_ones = torch.ones_like(x_data)
print("\nOnes Tensor:")
print(x_ones)

# Create random values with the same shape
# rand_like() generates random values
x_rand = torch.rand_like(x_data, dtype=torch.float)
print("\nRandom Tensor:")
print(x_rand)

# Create a 4x4 tensor containing ones
tensor = torch.ones(4, 4)
print("\nOnes Tensor:")
print(tensor)

# Create a 4x4 tensor containing zeros
tensor1 = torch.zeros(4, 4)
print("\nZeros Tensor:")
print(tensor1)

# Join two tensors row-wise
# torch.cat() combines tensors
tensor2 = torch.cat([tensor, tensor1], dim=0)
print("\nConcatenated Tensor:")
print(tensor2)

# Element-wise multiplication
print("\nMultiplication:")
print(tensor * tensor1)

# Transpose the tensor
print("\nTranspose:")
print(tensor.T)

# Add 3 to every element
# add_() updates the original tensor
tensor.add_(3)
print("\nAfter Adding 3:")
print(tensor)

# Create a tensor of five ones
t = torch.ones(5)
print("\nPyTorch Tensor:")
print(t)

# Convert PyTorch tensor into NumPy array
n = t.numpy()
print("\nNumPy Array:")
print(n)
print("Data Type:", type(n))



print("\nPyTorch Implementation Completed!")
```
