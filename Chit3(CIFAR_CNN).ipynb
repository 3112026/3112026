```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Input
from tensorflow.keras.optimizers import SGD
train = pd.read_csv(r"D:\Downloads\DL DATASET\cifar10\train.csv")
test = pd.read_csv(r"D:\Downloads\DL DATASET\cifar10\test.csv")

```

```python
X_train = train.drop('label', axis=1).values
y_train = train['label'].values

X_test = test.values
```

```python
# Normalize data
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
# c. Define Feedforward Neural Network
model = Sequential([
    Input(shape=(3072,)),
    Dense(512, activation='relu'),
    Dense(256, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
# d. Compile and train using SGD
model.compile(
    optimizer=SGD(learning_rate=0.01),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train, y_train,
    epochs=20,
    batch_size=32,
    validation_split=0.2
)
```

```python

# e. Evaluate the model
loss, accuracy = model.evaluate(X_train, y_train)
```

```python
# f. Plot training loss and accuracy
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(history.history['loss'])
plt.title("Training Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")

plt.subplot(1, 2, 2)
plt.plot(history.history['accuracy'])
plt.title("Training Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.show()
```

```python
# Display sample predictions
predictions = model.predict(X_train[:5])
predicted_labels = np.argmax(predictions, axis=1)

plt.figure(figsize=(10, 4))

for i in range(5):
    image = X_train[i].reshape(3, 32, 32).transpose(1, 2, 0)

    plt.subplot(1, 5, i + 1)
    plt.imshow(image)
    plt.title("Actual: " + str(y_train[i]) +
              "\nPredicted: " + str(predicted_labels[i]))
    plt.axis('off')

plt.tight_layout()
plt.show()
```
